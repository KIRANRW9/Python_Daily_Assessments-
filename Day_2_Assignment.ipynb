Q1. Wikipedia ia a free online encylopidia, created by volunteers around the world and hosted by Wikipedia Foundation.
Find the No of words in a given sentence.

```python
str="Wikipedia ia a free online encylopidia, created by volunteers around the world and hosted by Wikipedia Foundation"
no_of_words=str.split(" ")
print("No.of Words:",len(no_of_words))
```
