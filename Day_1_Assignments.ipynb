```python
import math

def  f(x):
    return x**2 + 3*x -4

values=[2,-1,1]

for value in values:
    result=(f(value)==0)
    print(result)


#quadratic equation
a,b,c=1,3,-4

disc=(b**2)-4*a*c

if disc>=0:
    r1=(-b+math.sqrt(disc))/(2*a)
    r2=(-b-math.sqrt(disc))/(2*a)
    print("The roots are:",r1,r2)
else:
    print("It has No real Roots")
```

```python
def points(wins,draws,loss):
    res=wins*3 + draws*1 + loss*0
    return res

values=input("Enter How many wins,draws,losses seperated by Comma").split(",")
wins,draws,losses=map(int,values)

total_points=points(wins,draws,losses)
print("Total Points",total_points)


```

```python
#Q4
values=input("Enter No.Of Apples, Oranges, Banana Seperated By Comma").split(",")
apples,orange,banana=map(int,values)
Total=apples+orange+banana

#Percentage
ApplePercentage=(apples/Total)*100
OrangePercentage=(orange/Total)*100
BananaPercentage=(banana/Total)*100

print("The Percentages are:")
print(f"Apple Percentage is {ApplePercentage:.2f}%")
print(f"Orange Percentage is {OrangePercentage:.2f}%")
print(f"Banana Percentage is {BananaPercentage:.2f}%")
```
