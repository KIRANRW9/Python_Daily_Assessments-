```python
Temperature=[34,40,29,33,42,37,39]
print("Orginal List",Temperature)

index_min=Temperature.index(min(Temperature))
index_max=Temperature.index(max(Temperature))
temp=Temperature[index_min]
Temperature[index_min]=Temperature[index_max]
Temperature[index_max]=temp

print("Converted List:",Temperature)
```

```python
def flatten_list(input_list):
    lst = []
    for item in input_list:
        if isinstance(item, list):
            lst.extend(flatten_list(item))  # Recursively flatten nested lists
        else:
            lst.append(item)
    return lst

tech_Companies = ['Qualcom', 'Google', 'Apple', ['Nvida',['a','b','c'], 'Cisco', 'Samsung']]
flattened_list = flatten_list(tech_Companies)
print(flattened_list)
```
